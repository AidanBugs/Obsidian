# 2
For the full derivation we simply replace the $\sigma(z^i)$ with $1-\sigma(z^i)$ thus:

$L =-\sum_{i=1}^N y^i \log (1-\sigma (z^i))+ (1-y^i)\log(\sigma (z^i))$

For $w_j$:

> $\frac{\delta L}{\delta w_j} =\sum_{i=1}^N -y^i \frac{1}{1-\sigma (z^i)}(-\sigma(z^i))(1-\sigma(z^i))x^i_j- (1-y^i)\frac{1}{\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))x^i_j$
> 
> $\frac{\delta L}{\delta w_j} =\sum_{i=1}^N y^i \sigma(z^i)x^i_j- (1-y^i)(1-\sigma(z^i))x^i_j$
> 
> $\frac{\delta L}{\delta w_j} =\sum_{i=1}^N y^i \sigma(z^i)x^i_j- (1-y^i-\sigma(z^i)+y^i\sigma(z^i))x^i_j$
> 
> $\frac{\delta L}{\delta w_j} =\sum_{i=1}^N (y^i-(1-\sigma(z^i)))x^i_j$ 

For $b$:

> $\frac{\delta L}{\delta b} =\sum_{i=1}^N -y^i \frac{1}{1-\sigma (z^i)}(-\sigma(z^i))(1-\sigma(z^i))- (1-y^i)\frac{1}{\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))$
> 
> $\frac{\delta L}{\delta b} =\sum_{i=1}^N y^i \sigma(z^i)- (1-y^i)(1-\sigma(z^i))$
> 
> $\frac{\delta L}{\delta b} =\sum_{i=1}^N y^i \sigma(z^i)- (1-y^i-\sigma(z^i)+y^i\sigma(z^i))$
> 
> $\frac{\delta L}{\delta b} =\sum_{i=1}^N (y^i-(1-\sigma(z^i)))$ 

Thus it follows that 

$$
w_j \leftarrow w_j - \eta \frac{\delta L}{\delta w_j}
$$
$$
b \leftarrow b - \eta \frac{\delta L}{\delta b}
$$

$L =\frac{\lambda}{2n}\sum_{i=1}^m w_i^2  -\sum_{i=1}^N y^i \log (\sigma (z^i))+ (1-y^i)\log(1-\sigma (z^i))$

For $w_j$:

>$\frac{\delta L}{\delta w_j} = \frac{\lambda w_j}{n} + \sum_{i=1}^N - y^i \frac{1}{\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))x^i_j+ (1-y^i)\frac{1}{1-\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))x^i_j$
>
>$\frac{\delta L}{\delta w_j} =\sum_{i=1}^N -y^i (1-\sigma(z^i))x^i_j + (1-y^i)(\sigma(z^i))x^i_j +\frac{\lambda w_j}{n}$
>
>$\frac{\delta L}{\delta w_j} =\sum_{i=1}^N (\sigma(z^i)-y^i)x^i_j+\frac{\lambda w_j}{n}$

For $b$:

>$\frac{\delta L}{\delta w_j} = - \sum_{i=1}^N y^i \frac{1}{\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))x^i_j+ (1-y^i)\frac{1}{1-\sigma (z^i)}\sigma(z^i)(1-\sigma(z^i))x^i_j$
>
>$\frac{\delta L}{\delta w_j} =\sum_{i=1}^N -y^i (1-\sigma(z^i))x^i_j + (1-y^i)(\sigma(z^i))x^i_j$
>
>$\frac{\delta L}{\delta w_j} =\sum_{i=1}^N (\sigma(z^i)-y^i)x^i_j$

Again it still applies that 

$$
w_j \leftarrow w_j - \eta \frac{\delta L}{\delta w_j}
$$
$$
b \leftarrow b - \eta \frac{\delta L}{\delta b}
$$

```python
#| echo: false

# Copied function from HW1 :)

import numpy as np
import matplotlib.pyplot as plt

def plot_points(points, bounds):
  x = np.linspace(bounds[0],bounds[1],100)
  y = np.linspace(bounds[0],bounds[1],100)
  X,Y = np.meshgrid(x,y)

  fig, ax = plt.subplots(figsize=(6, 4))
    
  class0_points = [(x, y) for x, y, c in points if c == 1]
  class1_points = [(x, y) for x, y, c in points if c == 2]
  
  if class0_points:
    class0_x, class0_y = zip(*class0_points)
    ax.scatter(class0_x, class0_y, c='red', s=100, edgecolors='black', linewidth=2, label='Class 1', marker='o')
  
  if class1_points:
    class1_x, class1_y = zip(*class1_points)
    ax.scatter(class1_x, class1_y, c='blue', s=100, edgecolors='black', linewidth=2, label='Class 2', marker='s')

  ax.set_xlabel('Input 1', fontsize=12)
  ax.set_ylabel('Input 2', fontsize=12)
  ax.set_xlim(bounds)
  ax.set_ylim(bounds)
  ax.grid(True, alpha=0.3)
  ax.axhline(0, color='black', linewidth=0.5)
  ax.axvline(0, color='black', linewidth=0.5)
  
  ax.legend(loc='upper right', fontsize=10)
  
  plt.tight_layout()

q4 = [
  [1,1,1],
  [1,2,1],
  [2,1,1],
  [0,0,2],
  [1,0,2],
  [0,1,2],
]

plot_points(q4, [-0.5,2.5])
```

The dataset is linearly seperable by the line $y=1.5-x$

```python
#| echo: false

# Yoinked from scikit learn docs with some slight modifications

from sklearn import svm
from sklearn.inspection import DecisionBoundaryDisplay


def plot_training_data_with_decision_boundary(
    kernel, ax=None, long_title=True, support_vectors=True
):
    # Train the SVC
    clf = svm.SVC(kernel=kernel, gamma=2).fit(X, y)

    # Settings for plotting
    if ax is None:
        _, ax = plt.subplots(figsize=(4, 3))
    x_min, x_max, y_min, y_max = -3, 3, -3, 3
    ax.set(xlim=(x_min, x_max), ylim=(y_min, y_max))

    # Plot decision boundary and margins
    common_params = {"estimator": clf, "X": X, "ax": ax}
    DecisionBoundaryDisplay.from_estimator(
        **common_params,
        response_method="predict",
        plot_method="pcolormesh",
        alpha=0.3,
    )
    DecisionBoundaryDisplay.from_estimator(
        **common_params,
        response_method="decision_function",
        plot_method="contour",
        levels=[-1, 0, 1],
        colors=["k", "k", "k"],
        linestyles=["--", "-", "--"],
    )

    if support_vectors:
        # Plot bigger circles around samples that serve as support vectors
        ax.scatter(
            clf.support_vectors_[:, 0],
            clf.support_vectors_[:, 1],
            s=150,
            facecolors="none",
            edgecolors="k",
        )

    # Plot samples by color and add legend
    ax.scatter(X[:, 0], X[:, 1], c=y, s=30, edgecolors="k")
    if long_title:
        ax.set_title(f" Decision boundaries of {kernel} kernel in SVC")
    else:
        ax.set_title(kernel)

    if ax is None:
        plt.show()

X = np.array([[2,0],[-2,0],[0,2],[0,-2],[1,0],[-1,0],[0,1],[0,-1]])
y = np.array([1,1,1,1,2,2,2,2])
plot_training_data_with_decision_boundary("rbf")
```

Gaussian kernel SVM can perfectly separate the two classes
